```python
# TODO: join observed and reference files by row_id. ???
val = val_ref
test = test_ref

# TODO: quantify and visualize label disagreement.
def disagreement_rate(df):
    return (df["income_observed"] != df["income_reference"]).mean()

def conditional_rates(df):
    ref_pos = df[df["income_reference"] == 1]
    ref_neg = df[df["income_reference"] == 0]
    return {
        "fn_rate_given_ref1": (ref_pos["income_observed"] == 0).mean(),  # ref=1, obs=0
        "fp_rate_given_ref0": (ref_neg["income_observed"] == 1).mean(),  # ref=0, obs=1
    }

rows = []
for name, df in [("Validation", val), ("Test", test)]:
    rates = conditional_rates(df)
    rows.append({
        "Split": name,
        "n": len(df),
        "Overall disagreement": disagreement_rate(df),
        "FN rate | ref=1": rates["fn_rate_given_ref1"],
        "FP rate | ref=0": rates["fp_rate_given_ref0"],
    })

disagreement_summary = pd.DataFrame(rows).set_index("Split")

# report-ready formatted view (percentages, 2 decimals)
disagreement_summary_pretty = disagreement_summary.copy()
for col in ["Overall disagreement", "FN rate | ref=1", "FP rate | ref=0"]:
    disagreement_summary_pretty[col] = (disagreement_summary_pretty[col] * 100).map("{:.2f}%".format)

disagreement_summary_pretty

# TODO: test the Week 2 hypotheses without reading the new metadata.
```
