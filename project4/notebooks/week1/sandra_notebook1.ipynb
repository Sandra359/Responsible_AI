```python
# get the derm7pt release v0 data from https://derm7pt.com/downloads/ and unzip it to ~/data/derm7pt/release_v0
DIR = os.path.expanduser("~/data/derm7pt/release_v0")
meta = pd.read_csv(os.path.join(DIR, "meta/meta.csv"))

# Same grouping as derm7pt's Derm7PtDatasetGroupInfrequent
def group_diagnosis(d):
    d = str(d).lower()
    if d == "basal cell carcinoma":   return "BCC"
    if "melanoma" in d:               return "MEL"
    if "nevus" in d:                  return "NEV"
    
    if d == "seborrheic keratosis":   return "SK"
    if d in {"dermatofibroma", "lentigo", "melanosis", "miscellaneous", "vascular lesion"}: return "MISC"
    return None                                            # unknown -> check

CONCEPT_MAPS = {
    "pigment_network":       {"absent": 0, "typical": 1, "atypical": 2},
    "blue_whitish_veil":     {"absent": 0, "present": 1},
    "vascular_structures":   {"absent": 0, "arborizing": 1, "comma": 1, "hairpin": 1, "within regression": 1,
                              "wreath": 1, "dotted": 2, "linear irregular": 2},
    "pigmentation":          {"absent": 0, "diffuse regular": 1, "localized regular": 1,
                              "diffuse irregular": 2, "localized irregular": 2},
    "streaks":               {"absent": 0, "regular": 1, "irregular": 2},
    "dots_and_globules":     {"absent": 0, "regular": 1, "irregular": 2},
    "regression_structures": {"absent": 0, "blue areas": 1, "white areas": 1, "combinations": 1},
}
DIAG_CLASSES = ["BCC", "NEV", "MEL", "MISC", "SK"]

df = meta.copy()
df["diag"] = df["diagnosis"].map(group_diagnosis)
df["y"] = df["diag"].map({c: i for i, c in enumerate(DIAG_CLASSES)})
for col, m in CONCEPT_MAPS.items():
    df[col + "_num"] = df[col].str.strip().str.lower().map(m)

# sanity check: nothing should be unmapped
unmapped = {c: df.loc[df[c + "_num"].isna(), c].unique().tolist() for c in CONCEPT_MAPS}
print("unmapped diagnoses:", df.loc[df["diag"].isna(), "diagnosis"].unique())
print("unmapped concepts:", {k: v for k, v in unmapped.items() if v})

# official split
idx = lambda name: pd.read_csv(os.path.join(DIR, f"meta/{name}_indexes.csv"))["indexes"].tolist()
df["split"] = "unused"
df.loc[idx("train"), "split"] = "train"
df.loc[idx("valid"), "split"] = "valid"
df.loc[idx("test"),  "split"] = "test"

print(df["split"].value_counts())
print(pd.crosstab(df["diag"], df["split"]))
df[["case_num", "clinic", "derm", "diag"] + [c + "_num" for c in CONCEPT_MAPS]].head()
```

# Model performance

```python
import numpy as np
# Y: heltal fra kolonnen y, som du allerede har i df
Y = df["y"].values.astype(np.int64)

# C: 7 binære concepts (1 = mistænkelig variant), samme regler som LBK
CONCEPTS = {
    'pigment_network_num':       [2],
    'blue_whitish_veil_num':     [1],
    'vascular_structures_num':   [2],
    'pigmentation_num':          [2],
    'streaks_num':               [2],
    'dots_and_globules_num':     [2],
    'regression_structures_num': [1],
}
C = np.stack([df[col].isin(vals).values for col, vals in CONCEPTS.items()], axis=1).astype(np.float32)
weights = np.array([2, 2, 2, 1, 1, 1, 1])
print('score reproduceret:', (C @ weights == df['seven_point_score'].values).mean())

# split_idx: samme som før
split_idx = {s: pd.read_csv(os.path.join(DIR, 'meta', f'{s}_indexes.csv'))['indexes'].values
             for s in ['train', 'valid', 'test']}

CLASS_NAMES = DIAG_CLASSES
CONCEPT_NAMES = list(CONCEPTS)
```

```python
import torch, torchvision
from torchvision import transforms
from PIL import Image
from tqdm import tqdm

FEAT_FILE = 'features_derm_resnet50.npy'

backbone = torchvision.models.resnet50(weights=torchvision.models.ResNet50_Weights.IMAGENET1K_V2)
backbone.fc = torch.nn.Identity()          # 2048-d feature vector per image
backbone.eval()

preprocess = transforms.Compose([
    transforms.Resize(256), transforms.CenterCrop(224), transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])])

def load(i):
    img = np.asarray(Image.open(os.path.join(DIR, 'images', df.iloc[i]['derm'])).convert('RGB'))
    img = img[25:-25, 25:-25, :]          # fjern sort kant, samme som derm7pt (crop_amount=25)
    return preprocess(Image.fromarray(img))


if os.path.exists(FEAT_FILE):
    X = np.load(FEAT_FILE)
else:
    feats, n, bs = [], len(df), 32
    with torch.no_grad():
        for s in tqdm(range(0, n, bs)):
            batch = torch.stack([load(i) for i in range(s, min(s + bs, n))])
            feats.append(backbone(batch).numpy())
    X = np.concatenate(feats)
    np.save(FEAT_FILE, X)

print(X.shape)   # forventet (1011, 2048)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

# 1) Overall test performance per model
results = pd.DataFrame({name: cbm.evaluate(m, D, 'te') for name, m in models.items()}).T
display(results.round(3))

# 2) Per-concept test AUC per regime
concept_models = {k: v for k, v in models.items() if not k.startswith('black')}
display(pd.DataFrame({k: cbm.per_concept_auc(m, D, CONCEPT_NAMES) for k, m in concept_models.items()}).round(3))

# 3) Confusion matrices
fig, axes = plt.subplots(1, len(concept_models), figsize=(5 * len(concept_models), 4))
for ax, (name, m) in zip(axes, concept_models.items()):
    with torch.no_grad():
        y_pred = m(D['Xte'])[2].argmax(1).numpy()
    ConfusionMatrixDisplay.from_predictions(D['Yte'].numpy(), y_pred, display_labels=CLASS_NAMES,
                                            ax=ax, colorbar=False)
    ax.set_title(name)
plt.tight_layout(); plt.show()
```

```python
# Test metrics for EVERY joint lambda (the one selected on validation is only one of them)
joint_test = pd.DataFrame({f'joint (λ={lam})': cbm.evaluate(m, D, 'te') for lam, m in joint_runs.items()}).T

# Balanced accuracy when f is fed the TRUE test concepts (the "oracle" check)
oracle = {}
for lam, m in joint_runs.items():
    with torch.no_grad():
        pred = m.f(D['Cte']).argmax(1).numpy()
    oracle[f'joint (λ={lam})'] = cbm.label_metrics(D['Yte'].numpy(), pred)['bal_acc']
joint_test['bal_acc_true_concepts'] = pd.Series(oracle)
display(joint_test[['acc', 'bal_acc', 'macro_f1', 'concept_acc', 'concept_auc', 'bal_acc_true_concepts']].round(3))
```

```python
rng = np.random.default_rng(0)
Cte = D['Cte'].numpy().copy()
noisy = Cte.copy()
flip = rng.random(Cte.shape) < 0.10          # 10 % fake annotation errors
noisy[flip] = 1 - noisy[flip]

for name, m in concept_models.items():
    with torch.no_grad():
        p = torch.sigmoid(m.g(D['Xte'])).numpy()
    disagree = np.abs(p - noisy)              # how much the model disagrees with the (noisy) annotation
    k = flip.sum()
    top = np.argsort(disagree.ravel())[::-1][:k]          # flag as many cells as there are errors
    precision = flip.ravel()[top].mean()
    print(f"{name:20s} flagged-error rate {precision:.2f}  vs random {flip.mean():.2f}")
```

Fordi en fejl betyder, at markeringen ikke passer med billedet. Modellens bud er dens bedste gæt på, hvad der faktisk er på billedet. Derfor:

Er lægens markering rigtig, passer den med billedet, og modellen er som regel enig. Uenigheden er lav.
Er lægens markering forkert (en af dem, vi vendte), passer den ikke med billedet. Modellen ser stadig billedet, som det er, og bliver uenig. Uenigheden er høj.

Høj uenighed er altså et tegn på, at markeringen måske er forkert. Det er derfor, vi sorterer efter den.
